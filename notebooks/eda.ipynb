# Exploratory Data Analysis (EDA) for Brain Tumor Dataset

This notebook performs exploratory data analysis on the brain MRI images for tumor detection.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
from PIL import Image
from sklearn.model_selection import train_test_split

# Set style
sns.set_style('darkgrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# Load dataset paths
data_dir = '../data'
yes_dir = os.path.join(data_dir, 'yes')
no_dir = os.path.join(data_dir, 'no')

# Check if directories exist
print('Yes directory exists:', os.path.exists(yes_dir))
print('No directory exists:', os.path.exists(no_dir))
```

```python
# Function to load images
def load_images_from_folder(folder):
    images = []
    for filename in os.listdir(folder):
        img_path = os.path.join(folder, filename)
        if img_path.endswith(('.jpg', '.jpeg', '.png')):
            img = cv2.imread(img_path)
            if img is not None:
                img = cv2.resize(img, (128, 128))
                images.append(img)
    return images

# Load images
yes_images = load_images_from_folder(yes_dir)
no_images = load_images_from_folder(no_dir)

print(f'Number of tumor images: {len(yes_images)}')
print(f'Number of no-tumor images: {len(no_images)}')
```

```python
# Visualize sample images
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for i in range(5):
    axes[0, i].imshow(cv2.cvtColor(yes_images[i], cv2.COLOR_BGR2RGB))
    axes[0, i].set_title('Tumor')
    axes[0, i].axis('off')
    
    axes[1, i].imshow(cv2.cvtColor(no_images[i], cv2.COLOR_BGR2RGB))
    axes[1, i].set_title('No Tumor')
    axes[1, i].axis('off')
plt.show()
```

```python
# Image statistics
def image_stats(images, label):
    heights = [img.shape[0] for img in images]
    widths = [img.shape[1] for img in images]
    channels = [img.shape[2] for img in images]
    
    print(f'{label} Images:')
    print(f'  Average height: {np.mean(heights):.2f}')
    print(f'  Average width: {np.mean(widths):.2f}')
    print(f'  Channels: {channels[0]}')
    print(f'  Total images: {len(images)}')
    print()

image_stats(yes_images, 'Tumor')
image_stats(no_images, 'No Tumor')
```

```python
# Prepare data for ML
X = np.array(yes_images + no_images)
y = np.array([1] * len(yes_images) + [0] * len(no_images))

# Normalize
X = X / 255.0

# Flatten for ML models
X_flat = X.reshape(X.shape[0], -1)

print(f'X shape: {X.shape}')
print(f'y shape: {y.shape}')
print(f'X_flat shape: {X_flat.shape}')
```

```python
# Split data
X_train, X_test, y_train, y_test = train_test_split(X_flat, y, test_size=0.2, random_state=42)

print(f'Train set: {X_train.shape[0]} samples')
print(f'Test set: {X_test.shape[0]} samples')
```
